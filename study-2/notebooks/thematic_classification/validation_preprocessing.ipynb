```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
path = "../../datasets/classification_results_feb_2025_themes_validation_groundTruth.csv"

try:
    df = pd.read_csv(path, encoding="utf-8")
except UnicodeDecodeError:
    df = pd.read_csv(path, encoding="latin-1") 
df_ground_truth = []
for item in df.iterrows():
    themes = [item[1]["theme_1"], item[1]["theme_2"]]
    df_ground_truth.append({
        "abstract": item[1]["abstract"],
        "chosen_Themes": themes
    })
y = np.array([item['chosen_Themes'] for item in df_ground_truth])
train_val, test = train_test_split(
    df_ground_truth,
    test_size=0.2,
    random_state=42
)

# Step 2: Train (60%) / Validation (20%)
train, val = train_test_split(
    train_val,
    test_size=0.25,   # 0.25 of 0.8 = 0.2 total for validation
    random_state=42
)
train = pd.DataFrame(train)
val = pd.DataFrame(val)
test = pd.DataFrame(test)
with open("../../datasets/thematic_classification/gold_standard_train_themes.json", "w") as f:
    f.write(train.to_json(orient="records", lines=False))
with open("../../datasets/thematic_classification/gold_standard_val_themes.json", "w") as f:
    f.write(val.to_json(orient="records", lines=False))
with open("../../datasets/thematic_classification/gold_standard_test_themes.json", "w") as f:
    f.write(test.to_json(orient="records", lines=False))
```
